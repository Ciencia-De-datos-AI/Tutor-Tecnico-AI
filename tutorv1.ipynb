````python
# imports

from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
import ollama
````

````python
# constantes

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
````

````python
# Configura el entorno
load_dotenv()
openai = OpenAI()
````

````python
my_question = input("Por favor, introduce tu pregunta:")
````

````python
# prompts

system_prompt = "Eres un tutor técnico útil que responde preguntas sobre código Python, ingeniería de software, ciencia de datos y LLM, utiliza el metodo Feynman para dar la explicación, y agrega emojis relevantes con la explicacion"
user_prompt = "Por favor, da una explicación detallada de la siguiente pregunta: " + my_question
````

````python
# messages

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
````

````python
# Haz que gpt-4o-mini responda, con streaming
stream = openai.chat.completions.create(model=MODEL_GPT, messages=messages,stream=True)    
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    response = response.replace("```","").replace("markdown", "")
    update_display(Markdown(response), display_id=display_handle.display_id)
````

````python
# Haz que Llama 3.2 responda

response = ollama.chat(model=MODEL_LLAMA, messages=messages)
reply = response['message']['content']
display(Markdown(reply))
````
